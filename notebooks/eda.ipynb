```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path = '../data/sentiment_data.csv'
df = pd.read_csv(path)

df = df.rename(columns={
    "Comment": "text",
    "Sentiment": "label"
})
```

```python
df = df.drop(columns='Unnamed: 0')
```

```python
df.info()
```

```python
df["label"].value_counts(normalize=True)
```

```python
df["label"].value_counts().plot(kind="bar")
```

```python
label_map = {
    0: "negative",
    1: "neutral",
    2: "positive"
}

df["label"] = df["label"].map(label_map)
```

```python
df["label"].isna().sum()
```

```python
df = df.dropna()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
df.head()
df["label"].value_counts()
```

```python
df_sample = df.sample(300, random_state=42)

df_sample.to_csv("../data/dataset.csv", index=False)
```
